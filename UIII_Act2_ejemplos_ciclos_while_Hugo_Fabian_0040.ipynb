#### Ejemplo de ciclo while


```python
# 1. Contador Interactivo (Operadores Relacionales y Aritméticos)
# Este ejemplo simula el comportamiento de un ciclo "para" (for), pero utilizando un while para repetir un mensaje las veces que el usuario decida
# Solicitar el número de repeticiones
limite = int(input("¿Cuántas veces quieres saludar? "))
contador = 1

while contador <= limite:
    print(f"Hola número {contador}")
    contador += 1  # Operador aritmético de incremento
```

```python
# 2. Validación de Entrada (Operadores Lógicos y Relacionales)
# El ciclo while es ideal para obligar al usuario a ingresar datos válidos (en este caso, una calificación entre 0 y 10)
nota = float(input("Ingresa la calificación (0-10): "))

# Repetir mientras la nota sea menor a 0 O mayor a 10
while nota < 0 or nota > 10:
    print("Error: La calificación debe estar entre 0 y 10.")
    nota = float(input("Intenta de nuevo: "))

print(f"Calificación aceptada: {nota}")
```

```python
# 3. Suma Acumulativa con Centinela (Operadores Relacionales y break)
# Este programa suma números ingresados hasta que el usuario escribe un "valor centinela" (en este caso, el número 0)
suma = 0
print("Calculadora de suma (Ingresa 0 para finalizar)")

while True:  # Bucle infinito controlado manualmente
    numero = float(input("Introduce un número a sumar: "))

    if numero == 0:  # Operador relacional de igualdad
        break  # Interrumpe el ciclo inmediatamente

    suma += numero  # Acumulador aritmético
    print(f"Suma actual: {suma}")

print(f"Resultado final: {suma}")
```

```python
# 4. Recorrido de Cadenas (Operadores Relacionales y Funciones de Longitud)
# Aunque comúnmente se usan otros ciclos para texto, se puede recorrer una cadena con while usando el índice y la longitud
palabra = input("Escribe una palabra para deletrear: ")
indice = 0

while indice < len(palabra):
    caracter = palabra[indice]
    print(f"Letra en posición {indice}: {caracter}")
    indice += 1
```

```python
# 5. Cálculo de Potencia (Operadores Aritméticos y Relacionales)
# Este ejemplo calcula base exponente
# sin usar el operador de potencia de Python, solo mediante multiplicaciones repetidas
base = int(input("Introduce la base: "))
exponente = int(input("Introduce el exponente (positivo): "))
resultado = 1
contador = 0

while contador < exponente:
    resultado *= base  # Multiplicación acumulativa
    contador += 1

print(f"El resultado de {base} elevado a {exponente} es: {resultado}")
```
